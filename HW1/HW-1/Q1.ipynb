```python
from hazm import Stemmer
from hazm import Lemmatizer
from hazm import Normalizer
```

Stemming

```python
# normalization for using stemmer + stemming

normalizer = Normalizer()
stemmer = Stemmer()

words = [
    "دویدند",
    "برمی‌گردانند",
    "کتابهایشان",
    "نشسته‌اند",
    "بافتهایم",
    "آمده‌اید",
    "بهترین",
    "می‌رود"
]

normalized_words = [normalizer.normalize(word) for word in words]
stemmed_words = {word: stemmer.stem(normalized_word) for word, normalized_word in zip(words, normalized_words)}

for word, stem in stemmed_words.items():
    print(f"{word} → {stem}")
```

Lemmatization

```python
lemmatizer = Lemmatizer()

words = [
    "دویدند",
    "برمی‌گردانند",
    "کتابهایشان",
    "نشسته‌اند",
    "بافته‌ایم",
    "آمده‌اید",
    "بهترین",
    "می‌رود"
]

lemmatized_words = {word: lemmatizer.lemmatize(word) for word in words}

for word, lemma in lemmatized_words.items():
    print(f"{word} → {lemma}")
```
